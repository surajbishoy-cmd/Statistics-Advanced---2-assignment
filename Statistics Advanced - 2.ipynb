**Question 1: What is hypothesis testing in statistics?**
>Hypothesis testing in statistics is a statistical method used to make decisions or draw conclusions about a population based on sample data.

**Question 2:What is the null hypothesis, and how does it differ from the alternative hypothesis?**
>The null hypothesis is a statement that assumes there is no significant difference, effect, or relationship between variables. It represents the initial assumption that we test using sample data.

**Question 3: Explain the significance level in hypothesis testing and its role in deciding the outcome of a test.**
>Significance level is the probability of rejecting the null hypothesis when it is actually true. It represents the maximum acceptable risk of making a Type I error.

**Question 4: What are Type I and Type II errors? Give examples of each.**
>1. Type I Error (False Positive)
A Type I error occurs when we reject the null hypothesis even though it is actually true.
- Example: A medical test concludes that a healthy person has a disease when the person is actually healthy.
>2. Type II Error (False Negative)
A Type II error occurs when we fail to reject the null hypothesis even though it is actually false.
- Example: A medical test concludes that a person does not have a disease when the person actually has the disease.

**Question 5: What is the difference between a Z-test and a T-test? Explain when to use each.**

>*Z-test*:

Uses the standard normal distribution.
Used when population standard deviation is known.
Commonly used for large samples.
Example: Testing average product weight when population SD is known.

>*T-test*

Uses the Student's t-distribution.
Used when population standard deviation is unknown.
Often used for small samples.
Example: Testing average student marks when population SD is unknown.

>Conclusion: Use a Z-test when the population standard deviation is known and a T-test when it is unknown.

Question 6: Write a Python program to generate a binomial distribution with n=10 and p=0.5, then plot its histogram.


```python
import numpy as np
import matplotlib.pyplot as plt

# Generate binomial distribution
data = np.random.binomial(n=10, p=0.5, size=1000)

# Plot histogram
plt.hist(data, bins=np.arange(-0.5, 11.5, 1), edgecolor='black')

plt.xlabel("Number of Successes")
plt.ylabel("Frequency")
plt.title("Binomial Distribution (n=10, p=0.5)")
plt.show()
```

Question 7: Implement hypothesis testing using Z-statistics for a sample dataset in
Python. Show the Python code and interpret the results.
sample_data = [49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6,
50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5,
50.1, 50.3, 50.4, 50.5, 50.0, 50.7, 49.3, 49.8, 50.2, 50.9,
50.3, 50.4, 50.0, 49.7, 50.5, 49.9]

```python
import numpy as np
from scipy.stats import norm

sample_data = [
    49.1, 50.2, 51.0, 48.7, 50.5, 49.8, 50.3, 50.7, 50.2, 49.6,
    50.1, 49.9, 50.8, 50.4, 48.9, 50.6, 50.0, 49.7, 50.2, 49.5,
    50.1, 50.3, 50.4, 50.5, 50.0, 50.7, 49.3, 49.8, 50.2, 50.9,
    50.3, 50.4, 50.0, 49.7, 50.5, 49.9
]

# Hypotheses
mu = 50
sigma = 1.0  # Assumed known population standard deviation
alpha = 0.05

# Calculate sample statistics
n = len(sample_data)
sample_mean = np.mean(sample_data)

# Calculate Z-statistic
z = (sample_mean - mu) / (sigma / np.sqrt(n))

# Two-tailed p-value
p_value = 2 * norm.sf(abs(z))

print("Sample Mean:", round(sample_mean, 4))
print("Z-statistic:", round(z, 4))
print("P-value:", round(p_value, 4))

if p_value < alpha:
    print("Reject the null hypothesis")
else:
    print("Fail to reject the null hypothesis")
```

Question 8: Write a Python script to simulate data from a normal distribution and
calculate the 95% confidence interval for its mean. Plot the data using Matplotlib.


```python
import numpy as np
import matplotlib.pyplot as plt
from scipy import stats

# Generate normal distribution data
data = np.random.normal(loc=50, scale=10, size=100)

# Calculate 95% confidence interval
mean = np.mean(data)
se = stats.sem(data)
ci = stats.t.interval(0.95, df=len(data)-1,
                      loc=mean, scale=se)

print("Sample Mean:", round(mean, 2))
print("95% Confidence Interval:", ci)

# Plot histogram
plt.hist(data, bins=10, edgecolor='black')
plt.axvline(mean, linestyle='--', label='Mean')
plt.xlabel("Values")
plt.ylabel("Frequency")
plt.title("Normal Distribution")
plt.legend()
plt.show()
```

Question 9: Write a Python function to calculate the Z-scores from a dataset and
visualize the standardized data using a histogram. Explain what the Z-scores represent
in terms of standard deviations from the mean.


```python
import numpy as np
import matplotlib.pyplot as plt

def calculate_z_scores(data):
    mean = np.mean(data)
    std = np.std(data)
    z_scores = (data - mean) / std
    return z_scores

# Dataset
data = np.array([10, 20, 30, 40, 50, 60, 70])

# Calculate Z-scores
z_scores = calculate_z_scores(data)
print("Z-scores:", z_scores.round(2))

# Plot histogram
plt.hist(z_scores, bins=5, edgecolor='black')
plt.xlabel("Z-scores")
plt.ylabel("Frequency")
plt.title("Histogram of Standardized Data")
plt.show()
```
